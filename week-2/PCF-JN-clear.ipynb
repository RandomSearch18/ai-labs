# Polynomial curve fitting examples

This notebook shows how to use the numpy polyfit function to fit a polynomial to data

```python
import matplotlib.pyplot as mp
import numpy as np
x = np.linspace( -10 , 10 , 5 ) #Return evenly spaced numbers over a specified interval.
print(x)
```

## Linear, quadratic and cubic functions

```python
# Example with linear data
y = 2*x + 5
coeff = np.polyfit(x,y,2) #finds the coefficient of a 2nd order degree
xn = np.linspace(-20,20,100)
yn = np.poly1d(coeff) #create the polynomial with the coefficients 
mp.plot( xn,yn(xn),x,y,'o') #plot

#https://numpy.org/doc/stable/reference/generated/numpy.polyfit.html
#https://numpy.org/doc/stable/reference/generated/numpy.poly1d.html
```

```python
#Example with data from a parabolic function 
y = 2*x*x + 5
coeff = np.polyfit(x,y,2) #finds the coefficient of a 2nd order degree
yn = np.poly1d(coeff) #create the polynomial with the coefficients 
mp.plot( xn,yn(xn),x,y,'o') #plot 
```

```python
#Example with data from a third degree function 
y = 2*x*x*x + 5
coeff = np.polyfit(x,y,2) #finds the coefficient of a 2nd order degree
yn = np.poly1d(coeff) #create the polynomial with the coefficients 
mp.plot( xn,yn(xn),x,y,'o') #plot 
```

```python
#Example with data from a third degree function 
y = 2*x*x*x + 5
coeff = np.polyfit(x,y,3) #finds the coefficient of a 2nd order degree
yn = np.poly1d(coeff) #create the polynomial with the coefficients 
mp.plot( xn,yn(xn),x,y,'o') #plot 
```

## Sinusoid function with noise

```python
import math
SAMPLES = 40
x = np.linspace(-10, 10, SAMPLES)
ground_truth = np.sin(1*x)
y = ground_truth + np.random.normal(0, 0.1, SAMPLES)
print(len(x))
print(len(y))
coeff = np.polyfit(x,y,3) #finds the coefficient of a 2nd order degree
yn = np.poly1d(coeff) #create the polynomial with the coefficients 
mp.plot( xn,yn(xn),x,ground_truth,x,y,'o') #plot 
mp.grid()
# ( xn,yn(xn),x,y_ground,x,y,'o')
```

```python
error = np.sum(np.power(yn(xn) - y,2))
print(error)
```

```python
y = np.sin(2*np.pi*x) + np.random.normal(0,0.1,50)
y_ground = np.sin(2*math.pi*x)
coeff = np.polyfit(x,y,8) #finds the coefficient of a 2nd order degree
yn = np.poly1d(coeff) #create the polynomial with the coefficients 
mp.plot( xn,yn(xn),x,y,'o') #plot 
mp.grid()
```

```python
error = np.sum(np.power(yn(xn) - y,2))
print(error)
```

## Example of high degree polynomial on few data points
Note that polyfit provides a warning

```python
x = np.linspace( -1 , 1 , 10 ) #Return evenly spaced numbers over a specified interval.
x_dense = np.linspace(-1,1,100)
xn = np.linspace(-1,1,10)
y = np.sin(2*np.pi*x) + np.random.normal(0,0.2,10)
y_ground = np.sin(2*math.pi*x_dense)
coeff = np.polyfit(x,y,10) #finds the coefficient of a 2nd order degree
yn = np.poly1d(coeff) #create the polynomial with the coefficients 
mp.plot( x_dense,yn(x_dense),x_dense,y_ground,x,y,'o') #plot 
mp.grid()
```

Very small error on the test set

```python
error = np.sum(np.power(yn(xn) - y,2))
print(error)
```

Let's create a test set and check the error again

```python
y = np.sin(2*np.pi*x) + np.random.normal(0,0.2,10)
error = np.sum(np.power(yn(xn) - y,2))
print(error)
```

```python
mp.plot( x_dense,yn(x_dense),x_dense,y_ground,x,y,'o') #plot 
mp.grid()
```

We can observe that the overfitted polynomial that was fitting the data points very well in the training set, is not fitting anymore the new test data
