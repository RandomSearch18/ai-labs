### **Task 2**: Define a class to build a neural network with 2 hidden layers and a single output layer. The layers should have 512, 256 and 10 neurons, respectively.

```python
import torch
from torch import nn

accelerator = torch.accelerator.current_accelerator(check_available=True)
device_type = accelerator.type if accelerator else "cpu"
print(f"Using {device_type} device")

class BeautifulNeuralNetwork(nn.Module):
    def __init__(self):
        super().__init__()
        self.flatten = nn.Flatten()
        NEURONS_INPUT = 28*28 # 28 px * 28px
        NEURONS_LAYER_1 = 256
        NEURONS_LAYER_2 = int(NEURONS_LAYER_1 / 2)
        NEURONS_OUTPUT = 10 # Because there are 10 classes
        self.hidden_layers = nn.Sequential(
            # nn.Mish(),
            nn.Linear(NEURONS_INPUT, NEURONS_LAYER_1),
            nn.ReLU(),
            nn.Linear(NEURONS_LAYER_1, NEURONS_LAYER_2),
            nn.ReLU(),
            nn.Linear(NEURONS_LAYER_2, NEURONS_OUTPUT)
        )

    def forward(self, x):
        x = self.flatten(x)
        output = self.hidden_layers(x)
        return output

```

### **Task 3**: Write a function for training the network above on the MNIST dataset
